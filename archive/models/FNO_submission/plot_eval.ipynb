```python
from pathlib import Path

import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
```

```python
errors_path = Path('fno_errors.npz')
with np.load(errors_path) as errors:
    resolutions = [128, 256, 512]
    fno_means = [errors[f'err_{n}'].mean() for n in resolutions]
    identity_means = [errors[f'id_{n}'].mean() for n in resolutions]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
```

```python
x = np.arange(len(resolutions))
w = 0.35
axes[0].bar(x - w / 2, fno_means, w, label='FNO')
axes[0].bar(x + w / 2, identity_means, w, label='Identity')
axes[0].set_xticks(x)
axes[0].set_xticklabels([str(n) for n in resolutions])
axes[0].set_xlabel('Resolution')
axes[0].set_ylabel('Mean relative L² error')
axes[0].set_yscale('log')
axes[0].set_title('FNO vs identity baseline')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
```

```python
growth = [fno_means[1] / fno_means[0], fno_means[2] / fno_means[0]]
axes[1].bar(['128→256', '128→512'], growth, color=['C2', 'C3'])
axes[1].axhline(1.0, color='gray', linestyle='--', linewidth=1)
axes[1].set_ylabel('Growth ratio')
axes[1].set_title('FNO error growth relative to 128')
for i, value in enumerate(growth):
    axes[1].text(i, value, f' {value:.2f}×', ha='center', va='bottom')
axes[1].grid(True, alpha=0.3)
```

```python
plt.tight_layout()
output_path = Path('plots_eval.png')
plt.savefig(output_path, dpi=150)
print(f'Saved {output_path}')
```
