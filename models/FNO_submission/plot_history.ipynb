```python
from pathlib import Path

import numpy as np
import torch
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
```

```python
checkpoint_path = Path('fno_modes64_all48_seed0.pt')
checkpoint = torch.load(checkpoint_path, map_location='cpu', weights_only=True)
train = np.asarray(checkpoint['train_hist'], dtype=float)
epochs = np.arange(len(train))
final_loss = float(train[-1])
```

```python
# The final model was trained on all training examples; this checkpoint has no validation curve.
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
```

```python
axes[0].plot(epochs, train, color='C0', linewidth=1.5)
axes[0].axhline(final_loss, color='red', linestyle=':', linewidth=1,
                label=f'last epoch = {final_loss:.4g}')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Training relative L²')
axes[0].set_yscale('log')
axes[0].set_title('FNO training curve')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
```

```python
tail_start = max(0, len(train) - 100)
axes[1].plot(epochs[tail_start:], train[tail_start:], color='C1', linewidth=1.5)
axes[1].axhline(final_loss, color='red', linestyle=':', linewidth=1,
                label=f'last epoch = {final_loss:.4g}')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Training relative L²')
axes[1].set_yscale('log')
axes[1].set_title('Final 100 epochs')
axes[1].legend()
axes[1].grid(True, alpha=0.3)
```

```python
plt.tight_layout()
output_path = Path('plots_training.png')
plt.savefig(output_path, dpi=150)
print(f'Saved {output_path}')
```
